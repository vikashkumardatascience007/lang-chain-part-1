```python
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage
from langchain_community.chat_models import ChatOllama
import warnings
warnings.filterwarnings("ignore")
```

```python
llm = ChatOllama(
    model="mistral:latest",
    temperature=0.7
)
my_message=[
    SystemMessage(content='You are gen-z assitant who always  answer in fun way'),
    HumanMessage(content='Bro, tell me fun fact')
]

llm.invoke(my_message).content
```

```python
## dynamic prompt in langchain to do this use PromptTemplate

from langchain_core.prompts import PromptTemplate


user_input = input("Enter a topic for fun fact : ")

dynamic_prompt = PromptTemplate.from_template("Write a fun fact about {topic}")

ready_prompt = dynamic_prompt.invoke({"topic" : user_input})

llm.invoke(ready_prompt).content
```

```python
# prompt with AI tone enabling to do this use ChatPromptTemplate
from langchain_core.prompts import ChatPromptTemplate 

prompt_template = ChatPromptTemplate.from_messages([
    ("system","You are a {tone} assistant"),
    ("user","Write a fun fact about {topic}.")
])

user_input = input("Enter a topic: ")
user_tone = input("Enter a tone: ")

ready_prompt = prompt_template.invoke({"topic":user_input,"tone":user_tone})

llm.invoke(ready_prompt.messages)
```
